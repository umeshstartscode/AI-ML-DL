# GCD

```python
def gcd(m, n):
  cd = []
  for i in range(1, min(m,n) + 1):
    if m % i == 0 and n % i == 0:
      cd.append(i)
  return cd[-1]

def gcd2(m, n):
  for i in range(1, min(m,n) + 1):
    if m % i== 0 and n % i == 0:
      result = i
  return result




# Euclids theorem


def gcd3(m ,n):
  a , b = max(m, n) , min(m,n)
  if a % b == 0:
    return b
  else:
    return gcd(b, a%b)

```

```python
class Point:
  def __init__(self, x, y):
    self.a = x
    self.b = y

  def translate(self, deltax, deltay):
    self.a += deltax
    self.b += deltay

  def distance(self):
    import math
    d = math.sqrt(self.a*self.a + self.b*self.b)
    return d

  def __str__(self):
    return '(' + str(self.a) + ',' + str(self.b) + ')'

  def __add__(self, p):
    return(Point(self.a + p.a, self.b + p.b))



c = Point(1,8)
d = Point(3,5)
print(c+d)
```
