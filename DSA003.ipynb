```python
# Digit Extraction

def countdigits(N):
  digits = []
  num = N
  while num > 0:
    last_digit = num % 10
    digits.append(last_digit)
    num = num // 10
  return len(digits), digits

countdigits(102020030330)

```

```python
# GCD

def gcd(a, b):
  m, n = max(a, b) , min(a, b)
  if m % n == 0:
    return n
  else:
    return gcd(n, m % n)

```

```python
# All divisors
import math as m

def divisors(c):
  divisor =[]
  for z in range(1, (int(m.sqrt(c))+1)):
    if c % z == 0 and c != z*z:
      divisor.append(z)
      divisor.append(c//z)
    elif c%z == 0:
      divisor.append(z)

  return sorted(divisor)

divisors(256)


```

```python
def prime(c):
  if divisors(c) == [1, c]:
    return True
  else:
    return False
```
