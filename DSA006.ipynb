```python
# Largest element

def largest_num(nums):
  largest = nums[0]
  i = 1
  while i < len(nums):
    if nums[i] > largest:
      largest = nums[i]
    i+=1
  return largest

largest_num([1,3,4,5,6,7,8,1,2,3,4,9,7,8,9,8,9,66,7,67,5,6,4])
```

```python
def second_largest(nums):
  n = len(nums)
  if n <=1:
    return -1
  largest = float('-inf')
  second = float("-inf")                                       # Brute Force -- Sort in ascending order and return -2th position
  for i in nums:
    if i > largest:
      second = largest
      largest = i
    elif i < largest and i > second:
      second = i
  return second



second_largest([1,3,4,5,6,7,8,1,2,3,4,9,7,8,9,8,9,66,7,67, 89,5,6,4])
```

```python
def left_rotatek(nums, k):
  n = len(nums)

  if n <= 1:
    return
  temp =[]

  k = k % n
  for i in range(k):
      temp.append(nums[-1])
      nums.pop()

  for j in range(len(temp)):
      nums.insert(0, temp[j])
  return nums
left_rotatek([1,2,3,4,5], 1000)

# brute force
```

```python
#optimal
def rotatebyk(nums, k):
  n = len(nums)

  if n<=1:
    return nums

  k = k % n

  left = 0
  right = n-1
  while left < right:
    nums[left], nums[right] = nums[right], nums[left]
    left += 1
    right -= 1

  le = 0
  ri = k-1
  while le < ri:
    nums[le], nums[ri] = nums[ri], nums[le]
    le += 1
    ri -= 1

  l = k
  r = n-1
  while l<r:
    nums[l], nums[r] = nums[r], nums[l]
    l += 1
    r -= 1

  return nums

rotatebyk([1,2,3,4,5,6,7], 12)
```
