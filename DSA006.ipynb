```python
# Largest element

def largest_num(nums):
  largest = nums[0]
  i = 1
  while i < len(nums):
    if nums[i] > largest:
      largest = nums[i]
    i+=1
  return largest

largest_num([1,3,4,5,6,7,8,1,2,3,4,9,7,8,9,8,9,66,7,67,5,6,4])
```

```python
def second_largest(nums):
  n = len(nums)
  if n <=1:
    return -1
  largest = float('-inf')
  second = float("-inf")                                       # Brute Force -- Sort in ascending order and return -2th position
  for i in nums:
    if i > largest:
      second = largest
      largest = i
    elif i < largest and i > second:
      second = i
  return second



second_largest([1,3,4,5,6,7,8,1,2,3,4,9,7,8,9,8,9,66,7,67, 89,5,6,4])
```

```python
def second_smallest(nums):
  n = len(nums)
  if n <=1:
    return -1
  smallest = float('inf')
  second = float("inf")                                       # Brute Force -- Sort in ascending order and return -2th position
  for i in nums:
    if i < smallest:
      second = smallest
      smallest = i
    elif i > smallest and i < second:
      second = i
  return second

second_smallest([1,1.5,3,4,5,6,7,8,1,2,3,4,9,7,8,9,8,9,66,7,67, 89,5,6,4])
```
