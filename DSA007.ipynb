```python
def findmissing(nums):
  s = 0
  n = len(nums)
  for i in nums:
    s += i
  total = (n+2)*(n+1)/2

  return int(total - s)



findmissing([1,2,3,4,5,6,7,9])
```

```python
# 2 sum

def twoSum(nums, k):
  n = len(nums)
  hash_map = {}
  for i in range(n):
    diff = k - nums[i]
    if diff in hash_map:
      return [hash_map[diff], i]
    else:
      hash_map[i] = i

# optimal


```

```python
# Linear search - easy


def linear_Search(nums, k):
  n = len(nums)
  for i in range(n):
    if nums[i] == k:
      return k
  return -1
```

```python
def unionofarrays(num1, num2):
  dic = {}
  union = []
  for i in num1:
    if i not in dic:
      dic[i] = 1
  for j in num2:
    if j not in dic:
      dic[j] = 1
  for key in dic:
    union.append(key)
  return union


# or use a set

def union2(num1, num2):
  seen = set()
  for x in num1:
    seen.add(x)

  for y in num2:
    seen.add(y)

  return list(seen)



```

```python
# union of sorted arrays

def union_sorted(num1, num2):

  i = 0
  j = 0
  n = len(num1)
  m = len(num2)
  res = []


  while i < n and j < m:
    if num1[i] < num2[j]:
      if len(res) == 0 or res[-1] != num1[i]:
        res.append(num1[i])
      i += 1
    elif num1[i] > num2[j]:
      if len(res) == 0 or res[-1] != num2[j]:
        res.append(num2[j])
      j += 1
    else:
      if len(res) == 0 or res[-1] != num1[i]:
        res.append(num1[i])
      i += 1
      j += 1


  while i < n:
    if len(res) == 0 or num1[i] != res[-1]:
      res.append(num1[i])
    i += 1
  while j < m:
    if len(res) == 0 or num2[j] != res[-1]:
      res.append(num2[j])
    j += 1

  return res


# A two pointer optimal solution for union of two sorted arrays, checks through each element of both lists, moves pointers accordingly

```

```python
def longestsubarray(nums, k):

  n = len(nums)
  max_len = 0

  for i in range(n):
    total = 0

    for j in range(i, n):
      total += nums[j]

      if total == k:
        if max_len < (j - i + 1):
          max_len = (j - i + 1)

      elif total > k:
        break

  return max_len

# brute force

```
