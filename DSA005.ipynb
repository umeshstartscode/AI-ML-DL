# Sorting

```python
# Selection Sort

def selection_sort(nums):
  n = len(nums)


  for i in range(0, n):
    min_index = i
    for j in range(i + 1, n):
      if nums[j] < nums[min_index]:
        min_index = j

    nums[i], nums[min_index] = nums[min_index] , nums[i]
  return nums

selection_sort([1, 42,50, 50, 50,6,7, 8])
```

```python
# For descending order

def reverse_sel(nums):
  n = len(nums)
  for i in range(0, n):
    max_ind = i
    for j in range(i+1, n):
      if nums[j] > nums[max_ind]:
        max_ind = j
    nums[i], nums[max_ind] = nums[max_ind], nums[i]
  return nums

reverse_sel([4,6,6 ,7,3,4,11,29,0])
```

```python
# Bubble Sort

def bubble_sort(nums):
  n = len(nums)
  for j in range(n-1, -1, -1):
    for i in range(0, j):
      if nums[i] > nums[i+1]:
        nums[i+1], nums[i] = nums[i], nums[i+1]
  return nums

bubble_sort([4,6,6 ,7,3,4,11,29,0])


```

```python
# Descending

def reverse_bubble(nums):
  n = len(nums)
  for j in range(n-1, -1 , -1):
    for i in range(0, j):
      if nums[i] < nums[i+1]:
        nums[i], nums[i+1] = nums[i+1], nums[i]
  return nums

reverse_bubble([4,6,6 ,7,3,4,11,29,0])
```

```python
# Insertion Sorting

def insertion_sort(nums):
  n= len(nums)
  for i in range(0, n):
    j = i -1
    while j >= 0 and nums[j] > nums[j+1]:
      nums[j], nums[j+1] = nums[j+1], nums[j]
      j -= 1
  return nums

insertion_sort([4,6,6,7,3,4,11,29,0])
```

```python
def merge(left, right):
  l = len(left)
  r = len(right)
  i = 0
  j = 0
  result = []
  while i < l and j < r:
    if left[i] < right[j]:
      result.append(left[i])
      i += 1
    else:
      result.append(right[j])
      j += 1

  result += left[i:]
  result += right[j:]

  return result



def merge_sort(nums):
  if len(nums) <= 1:
    return nums

  mid = len(nums) // 2
  left = merge_sort(nums[:mid])
  right = merge_sort(nums[mid:])

  return merge(left, right)

```
