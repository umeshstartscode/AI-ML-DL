```python
# Factorial

def Factorial(N, tot = 1):

  if N == 0 or N == 1:
    return tot
  else:
    tot = tot * N
    return Factorial(N-1, tot)

Factorial(5)
```

```python
def reverse_arr(arr):
  rev_arr = []
  q =len(arr)
  for i in range(q):
    rev_arr.append(arr[-1-i])
  return rev_arr



```

```python
def reverse_array(arr, s= 0):

  r = len(arr)
  mid = (r // 2)
  if s >= mid:
    return arr

  arr[s], arr[-1-s] = arr[-1-s], arr[s]


  return reverse_array(arr, s+1)



reverse_array([1,2,3,4])
```

```python
def palindrome(b):
  b = b.lower()
  left, right = 0, len(b)-1

  while left <= right:
    if b[left] != b[right]:
        return False
    left+=1
    right-=1
  if left > right:
    return True


palindrome("Anna")
```

```python
def fib(n):
  if n == 0 or n == 1:
    return n
  return fib(n-1) + fib(n-2)
```

# Hashing

```python
nums = [1,3,4,5,3,4,24,5,3,5,78,5,0]
freq_map = {}
for num in nums:
  if num in freq_map:
    freq_map[num] += 1
  else:
    freq_map[num] = 1


#-----------------------------

for key, values in freq_map.items():
  print(key, "-->", values)
```

```python
# Method 2

hash_map = {}
o = len(nums)
for l in range(0, o):
  hash_map[nums[l]] = hash_map.get(nums[l], 0) + 1

print(hash_map)
```
